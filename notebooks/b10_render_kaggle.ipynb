```python
import shutil, zipfile
from pathlib import Path
import pandas as pd

manifest = pd.read_csv('manifests/attacker_test.csv')
missing = [p for p in manifest['path'] if not (Path('/tmp/b10') / p).exists()]
masks = sorted(Path('/tmp/b10/images').glob('*_mask.png'))
with zipfile.ZipFile('/kaggle/working/ocr_masks_attacker_test.zip', 'w') as z:
    for m in masks:
        z.write(m, m.name)
packed = 0
with zipfile.ZipFile('/kaggle/working/images_attacker_test.zip', 'w', zipfile.ZIP_STORED) as z:
    for p in manifest['path']:
        z.write(Path('/tmp/b10') / p, p)
        packed += 1
shutil.copy('manifests/attacker_test.csv', '/kaggle/working/attacker_test.csv')
print(len(manifest), 'manifest rows,', len(missing), 'missing,', packed, 'packed,', len(masks), 'OCR masks')
print(manifest.groupby(['setting', 'defence']).size().to_string())
print('output:', {p.name: f'{p.stat().st_size / 1e6:.0f} MB' for p in sorted(Path('/kaggle/working').iterdir()) if p.is_file()})
print('OK' if not missing and same and packed == len(manifest) else 'PROBLEM')
```
